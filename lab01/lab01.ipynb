# **Bài 7:**

Việc thay dim=0 (từ trên xuống) thành dim=1 (từ trái sang phải) nghĩa thay đổi cách tính toán như:

dim= 0 sẽ tính trung bình của các cột (tính trung bình trên từng cột) cho ra 4 kết quả

dim = 1 sẽ tính trung bình của các hàng(tính trung bình theo từng hàng) cho ra 3 kết quả

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))


print("\nThay dim=0 thành dim=1")
print("\nMean:")
print(torch.mean(a, dim=1))
print("\nStandard deviation:")
print(torch.std(a, dim=1))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
```


# **Bài 8**

## Hàm $y = x^2$

### Tính thủ công
- **Hàm số:** $y = x^2$
- **Công thức đạo hàm:** $y' = \frac{dy}{dx} = 2x$
- **Tính tại $x = 3.6$:**
  $$y'(3.6) = 2 \times 3.6 = 7.2$$

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

y = 3 * x**2 + 2 * x + 1
y.backward()

print("Thay đổi lại công thức:")
print("x =", x.item())
print("y =", y.item())
print("x.grad =", x.grad.item())
```

# **Bài 9:**

Kết quả gần xấp xỉ 1,2,3


*   optimizer.zero_grad() (Xóa gradient cũ):Mặc định, PyTorch sẽ cộng dồn (accumulate) các gradient sau mỗi lần gọi backward().

*   mse.backward() (Tính gradient mới - Backpropagation):
Dựa trên hàm mất mát mse (Mean Squared Error), hàm này sẽ chạy ngược đồ thị tính toán để tìm ra đạo hàm riêng (gradient) của hàm loss đối với tất cả các tensor có requires_grad=True (ở đây là w). Kết quả đạo hàm được lưu tự động ngầm vào thuộc tính w.grad.

*   optimizer.step() (Cập nhật trọng số):
Dựa trên gradient vừa tính được ở bước backward(), bộ tối ưu hóa (ở đây là NAdam) sẽ tiến hành cập nhật các giá trị của tensor w theo công thức toán học của nó, kết hợp với tốc độ học (learning rate lr=0.01). Việc này đưa w dần về phía giá trị tối ưu [1, 2, 3].





```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
  optimizer.zero_grad()
  y_pred = x @ w
  mse = torch.mean(torch.square(y - y_pred))
  mse.backward()
  optimizer.step()
print("Learned coefficients:")
print(w)
```

# **Bài 10**:


```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
  y1 = A + B - 9
  y2 = C - D - 1
  y3 = A + C - 8
  y4 = B - D - 2
  sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
  optimizer.zero_grad()
  sqerr.backward()
  optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```
